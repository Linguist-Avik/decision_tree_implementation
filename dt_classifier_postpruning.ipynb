# Decision Tree Implementation - Post Pruning

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy
```

### Dataset - Iris Dataset

```python
from sklearn.datasets import load_iris

iris=load_iris()
```

```python
iris
```

```python
print(iris['DESCR'])
```

### Independent and Dependent features

```python
X = pd.DataFrame(iris['data'], columns=['sepal length', 'sepal width', 'petal length', 'petal width'])
y = iris['target']
```

### Train Test Split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=.20, random_state=42)
```

### Decision Tree Classifier

- Does not require standardization

```python
from sklearn.tree import DecisionTreeClassifier

tree_classifier = DecisionTreeClassifier()
```

### Model Training

```python
tree_classifier.fit(X_train, y_train)
```

### Decision Tree Visualization

```python
from sklearn import tree

plt.Figure(figsize=(25,15))
tree.plot_tree(tree_classifier, filled=True)
```

### Decision Tree with Max Depth = 2 Paramter

- This is Post Pruning Technique. Post Pruning is good for smaller dataset but for larger dataset we need to apply Pre Pruning

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn import tree


## DT Initialization with max_depth = 2
tree_classifier = DecisionTreeClassifier(max_depth=2)

## Model Training
tree_classifier.fit(X_train, y_train)

### Visualization
plt.Figure(figsize=(25,15))
tree.plot_tree(tree_classifier, filled=True)
```

### Model Prediction

```python
y_prediction = tree_classifier.predict(X_test)
```

### Metrics

```python
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score

print("Accuracy Score       :\n", accuracy_score(y_test, y_prediction))
print("Confusion Matrix     : \n", confusion_matrix(y_test, y_prediction))
print("Classification report: \n", classification_report(y_test, y_prediction))
```
